```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("clean_final_data.csv")

# Convert OrderDate into date format
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

# Display first 5 rows
df.head()
```

```
print("Rows and Columns:", df.shape)

```

```
print("\nColumn Names:")
print(df.columns.tolist())
```

```
print("\nData Types:")
print(df.dtypes)
```

```
print("\nMissing Values:")
print(df.isnull().sum())
```

```
df.describe()
```

# **Distribution of Sales**

```
plt.figure(figsize=(10, 5))

sns.histplot(df['Sales'], bins=50, kde=True)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

# **Customer Age Distribution**

```
plt.figure(figsize=(10, 5))

sns.histplot(df['Age'], bins=20, kde=True)

plt.title("Customer Age Distribution")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")

plt.show()
```

# **quantity distribution**

```
plt.figure(figsize=(10, 6))

sns.countplot(x='Quantity',data=df)
#sns.histplot(df['Quantity'], bins=20, kde=True)

plt.title("Quantity Distribution")
plt.xlabel("Quantity per Order")
plt.ylabel("Number of Orders")

plt.show()
```

# **Payment Method Distribution**(use count plot)

```
payment_counts = df['PaymentMethod'].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=payment_counts.index,
    y=payment_counts.values
)

plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")

plt.xticks(rotation=20)

plt.show()
```

# **Order Status Distribution**

```
status_counts = df['Status'].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=status_counts.index,
    y=status_counts.values
)

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")

plt.show()
```

# **Customer Segment Distribution**

```
segment_counts = df['CustomerSegment'].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=segment_counts.index,
    y=segment_counts.values
)

plt.title("Customer Segment Distribution")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers/Orders")

plt.show()
```

# **Product Category Distribution**

```
category_counts = df['Category'].value_counts()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)

plt.title("Product Category Distribution")
plt.xlabel("Number of Orders")
plt.ylabel("Product Category")

plt.show()
```

# **Category-wise Total Sales**

```
category_total_sales = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)

print(category_total_sales)
```

# **graph Category-wise Total Sales**

```
plt.figure(figsize=(10, 6))

category_total_sales.plot(kind='bar')

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

# **Category-wise Order Value**

```
category_order_value = (
    df.groupby('Category')['OrderValue']
    .sum()
    .sort_values(ascending=False)
)

print(category_order_value)
```

# **graph category-wise Order Value**

```
plt.figure(figsize=(10, 6))

category_order_value.plot(kind='bar')

plt.title("Category-wise Total Order Value")
plt.xlabel("Category")
plt.ylabel("Total Order Value")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

# **Calculate quantity sold by category**

```
category_quantity = (
    df.groupby('Category')['Quantity']
    .sum()
    .sort_values(ascending=False)
)

print(category_quantity)
```

# **Graph Category-wise  quantity sold**

```
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_quantity.values,
    y=category_quantity.index
)

plt.title("Category-wise Quantity Sold")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Product Category")

plt.show()
```

# **Daily Sales Trend**

```
daily_sales = df.groupby('OrderDate')['Sales'].sum().reset_index()

plt.figure(figsize=(14, 5))

plt.plot(
    daily_sales['OrderDate'],
    daily_sales['Sales']
)

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## **Discount vs Sales**

```
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Discount',
    y='Sales',
    alpha=0.5
)

plt.title("Discount vs Sales")
plt.xlabel("Discount (%)")
plt.ylabel("Sales")

plt.show()
```

# **Sales Distribution by Category**

```
category_sales = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.values,
    y=category_sales.index
)

plt.title("Sales Distribution by Category")
plt.xlabel("Total Sales")
plt.ylabel("Category")

plt.show()
```

# **summary table**

```
summary = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Order_Value=('OrderValue', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('OrderID', 'count')
).sort_values('Total_Sales', ascending=False)

print(summary)
```

# **Complete dashboard-style analysis**

```
print("========== SALES ANALYSIS ==========")

print("\n1. Sales Statistics")
print(df['Sales'].describe())

print("\n2. Customer Age Statistics")
print(df['Age'].describe())

print("\n3. Quantity Distribution")
print(df['Quantity'].value_counts().sort_index())

print("\n4. Payment Method Distribution")
print(df['PaymentMethod'].value_counts())

print("\n5. Order Status Distribution")
print(df['Status'].value_counts())

print("\n6. Customer Segment Distribution")
print(df['CustomerSegment'].value_counts())

print("\n7. Product Category Distribution")
print(df['Category'].value_counts())

print("\n8. Category-wise Total Sales")
print(df.groupby('Category')['Sales'].sum().sort_values(ascending=False))

print("\n9. Category-wise Order Value")
print(df.groupby('Category')['OrderValue'].sum().sort_values(ascending=False))
```
